## Problem 1: Vector operations

Let \(u=(1,2,3)\) and \(v=(4,5,6)\). Calculate their dot product and Euclidean norms.

\[
u\cdot v=1(4)+2(5)+3(6)=32,
\qquad \|u\|_2=\sqrt{1^2+2^2+3^2}=\sqrt{14}.
\]

```python
import numpy as np

u = np.array([1, 2, 3], dtype=float)
v = np.array([4, 5, 6], dtype=float)

dot_product = np.dot(u, v)
norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)

print("Dot product:", dot_product)
print(f"||u||₂ = {norm_u:.6f}")
print(f"||v||₂ = {norm_v:.6f}")
assert np.isclose(dot_product, 32)
assert np.isclose(norm_u, np.sqrt(14))
```

## Problem 2: Solve a linear system

Solve

\[
2x+y=5,\qquad x-y=1.
\]

In matrix form, \(Ax=b\), where \(A=\begin{bmatrix}2&1\\1&-1\end{bmatrix}\) and \(b=\begin{bmatrix}5\\1\end{bmatrix}\).

```python
A = np.array([[2, 1], [1, -1]], dtype=float)
b = np.array([5, 1], dtype=float)

solution = np.linalg.solve(A, b)
residual = A @ solution - b

print("Solution [x, y]:", solution)
print("Residual:", residual)
assert np.allclose(A @ solution, b)
assert np.allclose(solution, [2, 1])
```

**Verification.** Substituting \(x=2,y=1\) gives \(2(2)+1=5\) and \(2-1=1\). The residual is approximately zero, up to floating-point precision.

## Problem 3: Check whether a matrix is singular

A square matrix is singular if it has no inverse. Examine the determinant and attempt to solve a system using a matrix whose rows are linearly dependent.

```python
B = np.array([[1, 2], [2, 4]], dtype=float)
det_B = np.linalg.det(B)

print(f"Determinant: {det_B:.8f}")
print("Rank:", np.linalg.matrix_rank(B))
assert np.isclose(det_B, 0.0, atol=1e-10)
assert np.linalg.matrix_rank(B) == 1
```

**Interpretation.** The second row is twice the first, so the rows are linearly dependent and the matrix has rank 1, not 2. Numerical determinant calculations can produce tiny nonzero floating-point values, so use a tolerance.
