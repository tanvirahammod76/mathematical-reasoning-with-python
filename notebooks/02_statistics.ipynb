## Problem 1: Mean and variance

For the data \(2,4,4,4,5,5,7,9\), calculate the mean, population variance, and sample variance.

The population variance divides the sum of squared deviations by \(n\). The sample variance divides by \(n-1\), an adjustment commonly used to estimate population variance from a sample.

```python
import numpy as np

data = np.array([2, 4, 4, 4, 5, 5, 7, 9], dtype=float)
mean = data.mean()
population_variance = data.var(ddof=0)
sample_variance = data.var(ddof=1)

# Independent manual calculation of population variance
manual_population_variance = sum((x - mean) ** 2 for x in data) / len(data)

print(f"Mean: {mean:.3f}")
print(f"Population variance: {population_variance:.3f}")
print(f"Sample variance: {sample_variance:.3f}")
assert np.isclose(population_variance, manual_population_variance)
```

**Interpretation.** The two variance values differ because they use different denominators. Always state which definition you use.

## Problem 2: Standard error of a sample mean

Suppose observations are independent with population standard deviation \(\sigma=12\), and a sample contains \(n=36\) observations. The standard error of the sample mean is

\[
SE(\bar X)=\frac{\sigma}{\sqrt n}=\frac{12}{6}=2.
\]

This formula assumes independent observations with the stated finite standard deviation.

```python
sigma = 12
n = 36
standard_error = sigma / np.sqrt(n)
print(f"Standard error: {standard_error:.2f}")
assert np.isclose(standard_error, 2.0)
```

## Problem 3: Monte Carlo approximation

Estimate the probability of getting at least 4 heads in 6 fair coin tosses using simulation. Compare it with the exact binomial probability.

**Note:** A simulation is approximate and varies with the random seed and number of trials.

```python
from math import comb

rng = np.random.default_rng(2026)
n_trials = 200_000
n_tosses = 6

# True represents heads; each entry is an independent fair toss.
tosses = rng.integers(0, 2, size=(n_trials, n_tosses))
simulated_probability = np.mean(tosses.sum(axis=1) >= 4)
exact_probability = sum(comb(n_tosses, k) for k in range(4, 7)) / (2 ** n_tosses)

print(f"Simulated probability: {simulated_probability:.4f}")
print(f"Exact probability:     {exact_probability:.4f}")
print(f"Absolute error:        {abs(simulated_probability - exact_probability):.4f}")
```

**Verification.** The exact result is \(22/64=0.34375\). With many trials, the simulation will typically be close, but it is not guaranteed to equal the exact value.
