**Learning goals:** differentiate functions, verify derivatives numerically, and solve a simple optimization problem.

## Problem 1: Differentiate a polynomial

Let \(f(x)=x^3-2x+1\). By the power rule,

\[
f'(x)=3x^2-2.
\]

Evaluate the derivative at \(x=2\). The expected value is \(10\).

```python
import numpy as np
import sympy as sp

x = sp.symbols("x")
f = x**3 - 2*x + 1
derivative = sp.diff(f, x)
value_at_2 = derivative.subs(x, 2)

print("f(x) =", f)
print("f'(x) =", derivative)
print("f'(2) =", value_at_2)
assert value_at_2 == 10
```

## Problem 2: Verify with a central finite difference

For small \(h\), approximate the derivative by

\[
f'(x)\approx\frac{f(x+h)-f(x-h)}{2h}.
\]

This is a numerical approximation, not the symbolic derivative. Its accuracy depends on the step size and floating-point effects.

```python
f_numeric = sp.lambdify(x, f, "numpy")
x0 = 2.0
h = 1e-5
finite_difference = (f_numeric(x0 + h) - f_numeric(x0 - h)) / (2 * h)
exact_derivative = float(sp.diff(f, x).subs(x, x0))

print(f"Finite-difference approximation: {finite_difference:.8f}")
print(f"Exact derivative:                {exact_derivative:.8f}")
print(f"Absolute error:                  {abs(finite_difference - exact_derivative):.3e}")
assert abs(finite_difference - exact_derivative) < 1e-6
```

## Problem 3: Find the minimum of a quadratic

Consider \(g(x)=(x-3)^2+2\). Since the squared term is nonnegative, \(g(x)\geq2\), with equality at \(x=3\). Calculus gives the same result: \(g'(x)=2(x-3)=0\) at \(x=3\), and \(g''(x)=2>0\).

```python
g = (x - 3)**2 + 2
g_prime = sp.diff(g, x)
g_second = sp.diff(g_prime, x)
critical_points = sp.solve(sp.Eq(g_prime, 0), x)

print("g'(x) =", g_prime)
print("g''(x) =", g_second)
print("Critical points:", critical_points)
print("Minimum value:", g.subs(x, critical_points[0]))
assert critical_points == [3]
assert g.subs(x, 3) == 2
assert g_second > 0
```

**Interpretation.** The critical point is a minimum because the second derivative is positive. The algebraic form also directly shows the global minimum.
